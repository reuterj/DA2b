## Question 1. Processing text file (4 points)

The questions below ask you to summarize and calculate summary statistics for the mutual funds in 'HW4_fund_data.txt'. The cell below creates `lol`. Each element of `lol` contains
- fund name (string)
- calendar year (int)
- annual return (float)]

Note: Different funds can have different start and/or end dates. The earliest returns are for 1995. The most recent returns are for 2024. For 'Alpha Fund, Inc.', the first return is 1996 and the last return is 2022.

```python
f = open("HW4_fund_data.txt","r")
lines = f.readlines()
f.close()

lol = []
for line in lines:
    *name, year, ret = line.split('|')
    fullname = ','.join(name)
    year = int(year)
    ret  = float(ret)
    lol.append([fullname, year, ret])  
```
